# My first Quantum simulation

```python
import pennylane as qml

dev = qml.device("default.qubit", wires=2)

@qml.qnode(dev)
def bell_state():
    qml.Hadamard(wires=0)
    qml.CNOT(wires=[0, 1])
    return qml.sample(wires=[0, 1])

# Visual ASCII / Text representation
print(qml.draw(bell_state)())

# Interactive graphical circuit renderer (requires matplotlib)
# pip install matplotlib
fig, ax = qml.draw_mpl(bell_state)()
fig.show()
```

Ploting Measurement Probability Histograms

```python
import matplotlib.pyplot as plt
import numpy as np

# Draw 1000 measurement shots
samples = bell_state(shots=1000)

# Convert 2-bit measurement arrays to string labels (e.g., '00', '11')
bitstrings = ["".join(map(str, row)) for row in samples]
unique, counts = np.unique(bitstrings, return_counts=True)

# Plot bar chart
plt.figure(figsize=(6, 4))
plt.bar(unique, counts, color="royalblue", width=0.4)
plt.title("Bell State Measurement Outcomes (1000 Shots)")
plt.xlabel("Basis State")
plt.ylabel("Counts")
plt.ylim(0, 700)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```
