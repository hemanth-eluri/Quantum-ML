# My first Quantum simulation

```python
import pennylane as qml

# 1. Define a 2-qubit simulator device
dev = qml.device("default.qubit", wires=2, shots=1000)

# 2. Build the quantum circuit (QNode)
@qml.qnode(dev)
def bell_state():
    # Put qubit 0 into superposition: (|0> + |1>) / sqrt(2)
    qml.Hadamard(wires=0)
    
    # Entangle qubit 0 (control) and qubit 1 (target)
    qml.CNOT(wires=[0, 1])
    
    # Measure both qubits
    return qml.sample(wires=[0, 1])

# 3. Execute and view results
samples = bell_state()
print("Sample of first 10 measurements [Qubit 0, Qubit 1]:")
print(samples[:10])
```
